# Population and Sample in Statistics

## Population

- Population refers to the entire group or set of individuals, objects, or data points that you want to study or get information about.
- It includes every member of that group.
- The characteristics of a population (like the exact mean or standard deviation) are called **parameters**.
- **Examples:**  
  - All students in a school.  
  - Every book in a library.  
  - All voters in a city.  

## Sample

- A sample is a smaller subset taken from the population.  
- It is used when it's impractical or impossible to collect data from the entire population.  
- The characteristics of a sample (like the sample mean or sample standard deviation) are called **statistics**.  
- Samples are used to make estimates or inferences about the whole population.  
- **Examples:**  
  - 50 students selected randomly from a school.  
  - 100 books randomly checked in a library.  
  - 500 voters questioned in a city survey.  

---


### Why Use Samples?

- Populations can be too large or difficult to study completely.
- Sampling saves time and resources.
- Proper sampling methods allow us to make reliable conclusions about the population using sample data.

## 🔷 1. Probability Sampling Methods


### 🟢 a) Simple Random Sampling
- **Explanation:** Every student has an equal chance of being selected.
- **Example:** From a list of 500 students, randomly pick 50 using a lottery method or random number generator.

---


### 🟢 b) Systematic Sampling
- **Explanation:** Select every *k-th* student from a list.
- **Example:** Arrange all 500 students alphabetically. Pick every 10th student (10th, 20th, 30th...).

---

### 🟢 c) Stratified Sampling
- **Explanation:** Divide the population into subgroups (strata), then randomly sample from each.
- **Example:** Divide students into 3 courses — Science, Commerce, and Arts. Randomly pick 20 from each course.

### 🟢 d) Cluster Sampling
- **Explanation:** Divide population into clusters, then randomly choose entire clusters.
- **Example:** Out of 10 colleges in your city, randomly select 3 colleges. Survey **all students** in those colleges.

# 📊 Measures of Central Tendency

---

##  1. Definition

**Measures of Central Tendency** are statistical values that describe the **center or typical value** of a dataset.  
They indicate where most of the data points lie — representing the “average” value around which data is distributed.

The **three main measures** are:
1. **Mean**
2. **Median**
3. **Mode**

---

## 🔹 2. Mean (Arithmetic Average)

### 📘 Definition:
The **mean** is the sum of all data values divided by the number of observations.

$$
\bar{x} = \frac{\sum_{i=1}^{n} x_i}{n}
$$

### 📊 Example:
Dataset: [10, 20, 30, 40, 50]

$$
\bar{x} = \frac{10 + 20 + 30 + 40 + 50}{5} = 30
$$

---

## 🔹 3. Median (Middle Value)

### 📘 Definition:
The **median** is the middle value when data is arranged in ascending or descending order.

- If **n** is odd → middle value  
- If **n** is even → average of the two middle values

### 📊 Example:
Dataset: [5, 10, 15, 20, 25, 30]

Since 6 (even number of values):
$$
\text{Median} = \frac{15 + 20}{2} = 17.5
$$

---

## 🔹 4. Mode (Most Frequent Value)

### 📘 Definition:
The **mode** is the data value that occurs **most frequently**.

### 📊 Example:
Dataset: [2, 3, 3, 5, 7, 7, 7, 8, 10]

$$
\text{Mode} = 7
$$

#  Measures of Variability (Dispersion)

---

##  1. Definition

**Measures of Variability** (also called **Measures of Dispersion**) describe **how spread out or scattered** the data values are around the center (mean or median).

> In short, variability tells us **how consistent or inconsistent** the data is.

When variability is:
- **Low:** data points are close together (more consistent)
- **High:** data points are spread far apart (less consistent)

---

## 📚 2. Common Measures of Variability

| Measure | Description | Formula (Population) | Formula (Sample) |
|----------|--------------|----------------------|------------------|
| **Range** | Difference between the highest and lowest value | $$ R = X_{max} - X_{min} $$ | $$ R = X_{max} - X_{min} $$ |
| **Variance** | Average of squared deviations from the mean | $$ \sigma^2 = \frac{\sum (x_i - \mu)^2}{N} $$ | $$ s^2 = \frac{\sum (x_i - \bar{x})^2}{n - 1} $$ |
| **Standard Deviation (SD)** | Square root of variance | $$ \sigma = \sqrt{\frac{\sum (x_i - \mu)^2}{N}} $$ | $$ s = \sqrt{\frac{\sum (x_i - \bar{x})^2}{n - 1}} $$ |
| **Interquartile Range (IQR)** | Range between 75th and 25th percentile | $$ IQR = Q_3 - Q_1 $$ | $$ IQR = Q_3 - Q_1 $$ |

---


#  Range 

##  What is Range?
The **range** of a dataset is the **simplest measure of spread (variability)**.  

$$
\text{Range} = \text{Maximum Value} - \text{Minimum Value}
$$


---

##  Example with Dataset
Dataset:  
$
[12, 18, 25, 26, 30, 34, 40, 45, 50]
$

- **Minimum (min)** = 12  
- **Maximum (max)** = 50  

$$
\text{Range} = 50 - 12 = 38
$$



---

##  How to Interpret Range
`- A **small range** → all values are close together.  
- A **large range** → data is more spread out.  
- ⚠️ **Range depends only on 2 values (min & max)**, so it is very sensitive to **outliers**.  

**Example:**  
If one extra score **= 100** is added →  
$
\text{New Range} = 100 - 12 = 88
$
This makes the spread look huge, even though most values are still between 12 and 50.

---

---

##  Variance Formula (Simplified)

$$
\text{Variance} = \frac{1}{N} \sum (x_i - \mu)^2
$$

- $ x_i $ : each data value  
- $ \mu $: mean  
-  N : number of data points

 It's the **average of the squared differences** from the mean.

---

##  What is Standard Deviation?

**Standard Deviation (SD)** is the **square root of variance**.

> It tells you **how much the data varies from the mean** in the **original units** (like marks, rupees, etc.).





$$
\text{Standard Deviation} = \sqrt{\text{Variance}}
$$

---

---
## Example: score={10,20,30,40,50}
## 🔹 Step 1: Calculate the Mean (μ)

$$
\text{Mean} = \frac{10 + 20 + 30 + 40 + 50}{5} = \frac{150}{5} = 30
$$

 So, the average score is **30**.

---

## 🔹 Step 2: Find Deviations from the Mean

| Score (X) | Deviation (X − μ) | Squared Deviation \((X - μ)^2\) |
|-----------|--------------------|----------------------------------|
| 10        | 10 − 30 = -20       | (-20)² = 400                     |
| 20        | 20 − 30 = -10       | (-10)² = 100                     |
| 30        | 30 − 30 = 0         | 0² = 0                           |
| 40        | 40 − 30 = 10        | 10² = 100                        |
| 50        | 50 − 30 = 20        | 20² = 400                        |

---

## 🔹 Step 3: Calculate Variance (σ²)

$$
\text{Variance} = \frac{\sum (X - μ)^2}{N} = \frac{400 + 100 + 0 + 100 + 400}{5} = \frac{1000}{5} = 200
$$

📌 Variance tells us how **spread out** the data is (in squared units).

---

## 🔹 Step 4: Calculate Standard Deviation (σ)

$$
\text{Standard Deviation} = \sqrt{\text{Variance}} = \sqrt{200} ≈ 14.14
$$

📌 This means, on average, each score is about **14.14 points away** from the mean.

---
This way you’d see that:

The middle value (30) sits on the mean.

Values (10 and 50) are 20 away from the mean (biggest deviations).

Values (20 and 40) are 10 away from the mean.

 these deviations give you the variance = 200 and SD ≈ 14.14.


- **Mean** is the average score: 30  
- **Variance** shows the average of the squared differences from the mean: 200  
- **Standard Deviation** is the square root of variance: ≈ 14.14  


- A **low standard deviation** means scores are tightly clustered around the mean.
- A **high standard deviation** means the scores are more spread out.

# 📊 Finding Outliers Using IQR (Step-by-Step)

---

# 🔹 What is IQR?

The **Interquartile Range (IQR)** measures the spread of the middle 50% of data.

$$
IQR = Q_3 - Q_1
$$

Where:
- $Q_1$ = First Quartile (25th percentile)  
- $Q_2$ = Median (50th percentile)  
- $Q_3$ = Third Quartile (75th percentile)  

---

# 🔹 Outlier Formula

$$
\text{Lower Bound} = Q_1 - 1.5 \times IQR
$$

$$
\text{Upper Bound} = Q_3 + 1.5 \times IQR
$$

👉 Any value:
- $< \text{Lower Bound}$ → Outlier  
- $> \text{Upper Bound}$ → Outlier  

---

# ✅ CASE 1: ODD NUMBER OF OBSERVATIONS

## 📊 Dataset:
$1, 3, 5, 7, 9, 11, 13$

---

### Step 1: Find Median ($Q_2$)

Middle value:

$$
Q_2 = 7
$$

---

### Step 2: Split Data (Exclude Median)

- Lower half: $1, 3, 5$  
- Upper half: $9, 11, 13$  

---

### Step 3: Find $Q_1$ -find median for lower half and $Q_3$-find median for upper half

$$
Q_1 = 3, \quad Q_3 = 11
$$

---

### Step 4: Calculate IQR

$$
IQR = Q_3 - Q_1 = 11 - 3 = 8
$$

---

### Step 5: Find Bounds

$$
\text{Lower Bound} = 3 - (1.5 \times 8) = 3 - 12 = -9
$$

$$
\text{Upper Bound} = 11 + (1.5 \times 8) = 11 + 12 = 23
$$

---

### Step 6: Identify Outliers

No values outside $(-9, 23)$  
✅ No outliers  

---

# ✅ CASE 2: EVEN NUMBER OF OBSERVATIONS

## 📊 Dataset:
$2, 4, 6, 8, 10, 12, 14, 100$

---

### Step 1: Find Median ($Q_2$)

$$
Q_2 = \frac{8 + 10}{2} = 9
$$

---

### Step 2: Split Data

- Lower half: $2, 4, 6, 8$  
- Upper half: $10, 12, 14, 100$  

---

### Step 3: Find $Q_1$ and $Q_3$

$$
Q_1 = \frac{4 + 6}{2} = 5
$$

$$
Q_3 = \frac{12 + 14}{2} = 13
$$

---

### Step 4: Calculate IQR

$$
IQR = Q_3 - Q_1 = 13 - 5 = 8
$$

---

### Step 5: Find Bounds

$$
\text{Lower Bound} = 5 - (1.5 \times 8) = 5 - 12 = -7
$$

$$
\text{Upper Bound} = 13 + (1.5 \times 8) = 13 + 12 = 25
$$

---

### Step 6: Identify Outliers

$$
100 > 25
$$

❗ $100$ is an outlier  

---

# 🔥 Key Differences (Odd vs Even)

| Step | Odd Count | Even Count |
|------|----------|-----------|
| Median ($Q_2$) | Middle value | Average of two middle values |
| Split | Exclude median | Divide equally |
| $Q_1$, $Q_3$ | Median of halves | Median of halves |

---

# ⚡ Quick Memory Tips

- Always **sort the data first**  
- **Odd → Remove median before splitting**  
- **Even → Split equally**  
- Use **$1.5 \times IQR$ rule**  



# 🧠 Hypothesis Testing :Machine Bolt Diameter Test

---

## 🎯 Problem
A machine produces **bolts** with a **target diameter = 5 mm**.  
We want to check if the **average diameter has increased**.

---

# 🔹 Step-by-Step Summary

| 🪜 **Step** | 🧾 **Description** | 💬 **Explanation / Example** |
|:------------|:------------------|:-----------------------------|
| **1️⃣ State the claim** | Machine makes bolts with mean diameter **5 mm** | Target (μ₀) = **5 mm** |
| **2️⃣ Define hypotheses** | - **Null (H₀): μ = 5 mm**  <br> - **Alternative (H₁): μ > 5 mm** | We test if the mean **has increased** |
| **3️⃣ Collect sample data** | Measure some sample bolts (e.g., 10 pieces) | Example: 5.1, 5.0, 5.2, 5.3 … |
| **4️⃣ Calculate sample mean (x̄)** | Find the average of sample diameters | Suppose x̄ = **5.15 mm** |
| **5️⃣ Perform test** | Use **t-test** or **z-test** | Formula:<br>  t = (x̄ − μ₀) / (s / √n) |
| **6️⃣ Set significance level (α)** | Choose **α = 0.05** | 5% chance of error |
| **7️⃣ Compare & conclude** | If **p-value < α**, reject H₀ | Example: p = 0.02 → Mean has increased ✅ |

---

# 🧩 Example Result

| **Step** | **Result** |
|:-----------|:-------------|
| **Null Hypothesis (H₀)** | μ = 5 mm |
| **Alternative Hypothesis (H₁)** | μ > 5 mm |
| **Sample Mean (x̄)** | 5.15 mm |
| **p-value** | 0.02 |
| **Significance Level (α)** | 0.05 |
| **Conclusion** | ✅ **Reject H₀ → Mean diameter has increased** |

---

# ✨ Summary
✅ This is a **One-Tailed Test (Right Tail)**  
✅ We used a **t-test / z-test** to compare sample mean with target  
✅ If **p-value < 0.05**, the **average diameter increased**

#  1. **Independent t-test Example (Unpaired Groups)**

---

##  Scenario

We want to compare **marks of boys and girls** in a math test.

| **Group** | **Scores** |
|:-----------|:-------------|
| **Boys** | [55, 60, 65, 70, 75] |
| **Girls** | [65, 70, 72, 68, 74] |

---

##  Objective
Check if there is a **significant difference** between the **average marks** of boys and girls.

---

# 🔹 Step-by-Step Summary

| 🪜 **Step** | 🧾 **Description** | 💬 **Explanation / Example** |
|:------------|:------------------|:-----------------------------|
| **1️⃣ State the claim** | Compare mean marks of **boys** and **girls** | Two independent groups |
| **2️⃣ Define hypotheses** | - **H₀:** μ₁ = μ₂ (means are equal) <br> - **H₁:** μ₁ ≠ μ₂ (means are different) | We test if there is any difference |
| **3️⃣ Collect data** | Boys: [55, 60, 65, 70, 75] <br> Girls: [65, 70, 72, 68, 74] | Two separate samples |
| **4️⃣ Calculate sample means** | Boys avg = **65.0** <br> Girls avg = **69.8** | Girls have slightly higher marks |
| **5️⃣ Perform t-test** | Use **independent t-test formula** to compare means | t = -2.23 |
| **6️⃣ Set significance level (α)** | α = **0.05** | 5% chance of being wrong |
| **7️⃣ Compute p-value** | p = **0.056** | Compare this with α |
| **8️⃣ Draw conclusion** | Since **p (0.056) > α (0.05)** → ❌ **Fail to Reject H₀** | There is **no significant difference** between boys’ and girls’ marks |

# 🎓 2. **Dependent t-test Example (Paired Groups)**

---

## 📘 Scenario

We want to test if a **coaching class improved scores**.  
We take scores of the **same students** before and after the class.

| **Student** | **Before** | **After** |
|:-----------:|:----------:|:---------:|
| 1 | 50 | 55 |
| 2 | 55 | 60 |
| 3 | 52 | 58 |
| 4 | 60 | 65 |
| 5 | 58 | 64 |

**Before:** `[50, 55, 52, 60, 58]`  
**After:**  `[55, 60, 58, 65, 64]`

---

## 🎯 Objective
Check if the **coaching class** caused a **significant increase** in scores (same students measured twice).

---

# 🔹 Step-by-Step Summary

| 🪜 **Step** | 🧾 **Description** | 💬 **Explanation / Example** |
|:------------|:------------------|:-----------------------------|
| **1️⃣ State the claim** | Check if coaching **increased** scores | Same students before & after |
| **2️⃣ Define hypotheses** | - **H₀:** μ_d = 0 (mean difference = 0)  <br> - **H₁:** μ_d > 0 (mean difference > 0) | One-tailed test (we expect increase) |
| **3️⃣ Collect paired data** | Before and After lists (same students) | Before: [50,55,52,60,58] <br> After: [55,60,58,65,64] |
| **4️⃣ Compute differences** | d = After − Before | d = [5, 5, 6, 5, 6] |
| **5️⃣ Calculate mean of differences (d̄)** | d̄ = (5+5+6+5+6)/5 = **5.4** | Average improvement per student |
| **6️⃣ Calculate sd of differences (s_d)** | s_d ≈ **0.548** | Sample standard deviation of d |
| **7️⃣ Compute t-statistic** | t = d̄ / (s_d / √n) | t ≈ **22.05** (n = 5) |
| **8️⃣ Set significance level (α)** | α = **0.05** | 5% risk of error |
| **9️⃣ Compute p-value** | p < **0.0001** (very small) | Compare p with α |
| **🔟 Draw conclusion** | Since **p < α**, **Reject H₀** | ✅ **Coaching improved scores** (significant increase) |

# 🎓 3. **Chi-Square Test of Independence**

---

## 🎯 Situation
We want to find out whether **Gender** and **Preference for a Subject** are **related**.

---

## 📊 Observed Data

|            | **Science** | **Arts** | **Total** |
|:-----------|:------------:|:---------:|:----------:|
| **Boys**   | 30 | 10 | 40 |
| **Girls**  | 10 | 30 | 40 |
| **Total**  | 40 | 40 | 80 |

---

## 🎯 Objective
Check if **Gender** and **Subject Preference** are **independent** or **related**.

---

# 🔹 Step-by-Step Summary

| 🪜 **Step** | 🧾 **Description** | 💬 **Explanation / Example** |
|:------------|:------------------|:-----------------------------|
| **1️⃣ State the claim** | We test if **gender** and **subject preference** are **independent** | Boys & girls choose subjects differently? |
| **2️⃣ Define hypotheses** | - **H₀:** Gender and subject preference are **independent** <br> - **H₁:** Gender and subject preference are **not independent** | Two-way relationship test |
| **3️⃣ Write observed frequencies (O)** | Use the given data table | e.g., O₁₁ = 30, O₁₂ = 10, etc. |
| **4️⃣ Compute expected frequencies (E)** | E = (Row Total × Column Total) / Grand Total | For Boys–Science: (40×40)/80 = 20 |
| **5️⃣ Calculate all expected (E)** | Boys–Science = 20 <br> Boys–Arts = 20 <br> Girls–Science = 20 <br> Girls–Arts = 20 | Expected table: all = 20 |
| **6️⃣ Apply Chi-Square formula** | χ² = Σ((O−E)² / E) | = ((30−20)²/20) + ((10−20)²/20) + ((10−20)²/20) + ((30−20)²/20) |
| **7️⃣ Compute χ² value** | χ² = 10 + 5 + 5 + 10 = **30** | Calculated test statistic |
| **8️⃣ Find degrees of freedom (df)** | df = (r−1)(c−1) = (2−1)(2−1) = 1 | One degree of freedom |
| **9️⃣ Set significance level (α)** | α = **0.05** | 5% risk of error |
| **🔟 Compare with critical value or p-value** | For df=1, χ²₀.₀₅ = 3.84 | χ² = 30 > 3.84 |
| **✅ Draw conclusion** | Since χ² > 3.84 → **Reject H₀** | ✅ **Gender and subject preference are related** |
